## 3. 그래프 1 — 히스토그램 (40점)

**질문 3-1.** 어떤 수치형 변수의 분포를 보나요?

**[3-1]** (작성)

```python
col = "total_bill"  # 본인 데이터에 맞게 변경

plt.hist(df[col], bins=10, edgecolor="black")
plt.title("Distribution of " + col)
plt.xlabel(col)
plt.ylabel("count")
plt.show()
```

---
## 4. 그래프 2 — 막대 (groupby) (40점)

**질문 4-1.** 무엇을 기준으로 무엇을 비교하나요?

**[4-1]** (작성)


```python
group_col = "day"
value_col = "tip"

df.groupby(group_col)[value_col].mean().plot(kind="bar", color="steelblue")
plt.title(f"Mean {value_col} by {group_col}")
plt.xlabel(group_col)
plt.ylabel(f"mean {value_col}")
plt.show()
```

## 5. 그래프 3 — 그래픽 강조 (30점)
