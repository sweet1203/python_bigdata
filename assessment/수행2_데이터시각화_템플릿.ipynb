## 4. 그래프 2 — 막대 (groupby) (40점)

**질문 4-1.** 무엇을 기준으로 무엇을 비교하나요?

**[4-1]** (작성)

**그래프 2 의도:** 어떤 그룹을 비교하고, 어떤 평균이나 개수를 보려는지 적으세요.

(여기에 작성)


```python
group_col = "day"
value_col = "tip"

df.groupby(group_col)[value_col].mean().plot(kind="bar", color="steelblue")
plt.title(f"Mean {value_col} by {group_col}")
plt.xlabel(group_col)
plt.ylabel(f"mean {value_col}")
plt.show()
```

**그래프 2 실행 후 기록**

- **예시 코드에서 내가 바꾼 부분:** (group_col, value_col, 색 등)
- **내가 직접 실행한 결과:** (가장 크거나 작은 그룹, 차이가 보이는 지점 등)
