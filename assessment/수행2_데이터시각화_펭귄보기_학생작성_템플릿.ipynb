# A. 코드 보기 모음 — penguins 예시

이 구역은 **코드 사용 방법을 익히는 보기**입니다. 아래 코드들은 그대로 제출하는 답안이 아닙니다.

보기에서 확인할 것:

내가 실제로 작성할 코드는 아래 **B. 학생 작성 구역**에서 새로 입력합니다.


## 보기 0 — 데이터 불러오기와 기본 확인


```python
import seaborn as sns
import matplotlib.pyplot as plt

example_df = sns.load_dataset("penguins").dropna()

print("보기 데이터셋: penguins")
print("크기:", example_df.shape)
example_df.head()
```

## 보기 1 — 히스토그램: 수치형 변수의 분포 보기

`body_mass_g`처럼 숫자로 된 열이 어떤 범위에 많이 모여 있는지 확인합니다.


```python
plt.hist(example_df["body_mass_g"], bins=10, edgecolor="black", color="skyblue")
plt.title("Penguins body mass distribution")
plt.xlabel("body_mass_g")
plt.ylabel("count")
plt.show()
```

## 보기 2 — 막대그래프: 그룹별 평균 비교

`species`처럼 종류를 나타내는 열로 묶고, `body_mass_g` 평균을 비교합니다.


```python
penguin_mass = example_df.groupby("species")["body_mass_g"].mean()

penguin_mass.plot(kind="bar", color="steelblue")
plt.title("Mean body mass by penguin species")
plt.xlabel("species")
plt.ylabel("mean body_mass_g")
plt.show()
```

## 보기 3 — 산점도와 hue: 색으로 그룹 강조하기

두 수치형 변수의 관계를 보고, `hue`로 범주형 변수를 색으로 구분합니다.


```python
sns.scatterplot(
    data=example_df,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species"
)
plt.title("Bill length vs bill depth by species")
plt.xlabel("bill_length_mm")
plt.ylabel("bill_depth_mm")
plt.show()
```

## 보기 코드에서 기억할 점

- `df["열이름"]`: 그래프에 사용할 열 선택
- `groupby("기준열")`: 기준열로 데이터를 묶기
- `mean()`: 평균 계산
- `hue="열이름"`: 색으로 그룹 구분
- `plt.title()`, `plt.xlabel()`, `plt.ylabel()`: 그래프 설명 추가
- `plt.show()`: 그래프 출력
